```python
from geopy.geocoders import Nominatim
import time
from pprint import pprint

app = Nominatim(user_agent="tutorial")

location = app.geocode("Changi Airport").raw
pprint(location)
```

```python
import pandas as pd
location_df = pd.DataFrame(location)
location_df
```

```python
location['lat'], location['lon']
```

```python
import folium

m = folium.Map(location=[location['lat'], location['lon']], zoom_start=11)
folium.Marker([location['lat'], location['lon']], popup="Company").add_to(m)
m.save("sample_map.html")
```
