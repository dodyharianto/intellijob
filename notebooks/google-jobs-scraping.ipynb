```python
def scrape_google_jobs(query):
    search_params = {
        "engine": "google_jobs",
        "q": query,
        "hl": "en",
        "location": "Singapore",
        "api_key": SERPAPI_API_KEY
    }

    results = serpapi.search(**search_params)
    jobs_results = results["jobs_results"]

    next_page_tokens = 0
    while True:
        results = serpapi.search(**search_params)
        if 'job_results' not in results:
            print('No more job results. ❌')
            break
        next_page_tokens += 10
    jobs_results = results['jobs_results']
    return jobs_results
```
