```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()

response = client.responses.create(
    model="gpt-4o-mini",
    tools=[{
        "type": "web_search",
        "user_location": {
            "type": "approximate",
            "country": "SG",
            "city": "Singapore",
            "region": "Singapore",
        }
    }],
    input="Find me the brief details (including the SME/MNC, industry, company website, company's career website) of this company: Changi Airport",
)

print(response.output_text)
```
